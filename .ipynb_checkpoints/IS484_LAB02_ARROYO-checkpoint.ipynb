## Representación de un problema mediante estados

*Parte A:* Completar la ficha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| Estado | Descripción |
|---|---|
| *Estado inicial* | El repartidor se encuentra en el Almacén |
| *Estado objetivo* | El repartidor ha llegado a la Zona Norte |
| *Acciones posibles* | El repartidor se puede desplazar a cualquier punto que esté conectado directamente al punto en el que se encuentra |
| *Costo de cada acción* | El costo de cada acción (cada tramo recorrido) es el mismo para todos los desplazamientos |

```python
mapa = {
"Almacen": ["Centro", "Terminal"],
"Centro": ["Almacen", "ZonaNorte", "Mercado"],
"Terminal": ["Almacen", "Mercado"],
"Mercado": ["Centro", "Terminal", "ZonaNorte"],
"ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

## Implementación de BFS

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]] # cola de caminos, empieza con el estado inicial
    visitados = set()
    while cola:
        camino = cola.pop(0) # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]
        if nodo_actual == objetivo:
            return camino
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: crear un nuevo camino que sea "camino + [vecino]"
                # y agregarlo a la cola
                pass

    return None # no se encontro un camino
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

## 2. Vectores y matrices con NumPy
## Parte A: Representar cuatro observaciones de un dataset simple                                                                                     
## Parte B: Acceder a una la y a una columna especíca de la matriz

```python
import numpy as np  # importamos numpy para trabajar con vectores y matrices

# Cada observacion es un vector con 2 caracteristicas: temperatura y humedad
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# Juntamos las 4 observaciones en una sola matriz (dataset)
# cada FILA es una observacion, cada COLUMNA es una caracteristica
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)  # (4, 2) = 4 observaciones, 2 caracteristicas

print("Primera observacion (fila 0):", dataset[0])       # accedemos a una fila completa
print("Todas las temperaturas (columna 0):", dataset[:, 0])  # accedemos a una columna completa (todas las filas, columna 0)
```

## 3.Producto escalar y distancia euclidiana
## Parte A: Calcular el producto escalar entre dos observaciones:

```python
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
```

## Parte B: Calcular la distancia euclidiana entre dos observaciones

```python
# Formula manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)
# Con la funcion de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

## 4. Ejercicio Práctico de Laboratorio
## Parte A:Implementación de DFS

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]]
    visitados = set()
    while pila:
        camino = pila.pop() # sacar el ULTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]
        if nodo_actual == objetivo:
            return camino
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
        for vecino in grafo[nodo_actual]:
            nuevo_camino = camino + [vecino]
            pila.append(nuevo_camino)
            pass
    return None
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

## Parte B: Encontrar la observación más cercana

```python
nueva_observacion = np.array([20.5, 48])
distancias = []

for i, obs in enumerate(dataset):
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observacion {i}: {d}")

indice_mas_cercano = np.argmin(distancias)
print("La observacion mas parecida es la numero", indice_mas_cercano)
```

## Parte C: Reexión
## ¿por qué la distancia euclidiana es una forma razonable demedir qué tan parecidas son dos observaciones?
La distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones porque permite calcular qué tan cerca o lejos están sus valores. Si la distancia es pequeña, significa que las observaciones tienen características similares.
## En qué casos podría no ser la mejor opción (porejemplo, si las características tienen escalas muy distintas, como soles vs. años)?
Sin embargo, no siempre es la mejor opción. Por ejemplo, si las características tienen escalas muy diferentes, una variable podría influir mucho más que otra en el cálculo. En un caso como soles y años, los soles podrían dominar la distancia. Para solucionarlo, se puede realizar una normalización o estandarización de los datos antes de calcular la distancia.
